**Data Preprocessing **

```python
from sklearn.model_selection import train_test_split
skew_cols          = ['Study_Hours']
other_numeric_cols = ['Age', 'Avg_Daily_Usage_Hours', 'Daily_Unlocks',
                      'Physical_Activity_Hours', 'Sleep_Hours_Per_Night']

normal_cols     = ['Gender','Academic_Level','Most_Used_Platform','Purpose_Of_Use','Grouped_columns']

ordinal_cols = ['Stress_Level']

feature_cols = skew_cols + other_numeric_cols + normal_cols + ordinal_cols
x = students[feature_cols]
y = students['Mental_Health_Score']
X_train, X_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import FunctionTransformer , StandardScaler, OrdinalEncoder,OneHotEncoder
from sklearn.compose import ColumnTransformer
#1 for Skewness pipeline
skewed_pipline = Pipeline(steps=[('log_transform',FunctionTransformer(np.log1p)),('scale',StandardScaler())

                                 ])


#2 numeric columns
plain_numeric_cols= Pipeline(steps = [('scale',StandardScaler())
                  ])



#4 ordinal coloumns
ordinal_pipeline = Pipeline(steps=[('encode',OrdinalEncoder(categories=[['Low','Medium','High','Very High' ]]))
                ])

#3 normal columns
normal_pipeline = Pipeline(steps=[('encode',OneHotEncoder(handle_unknown='ignore'))])

prepocessor = ColumnTransformer(transformers=[('skewed_cols', skewed_pipline ,skew_cols),
                                                   ('Plain_Numeric',plain_numeric_cols , other_numeric_cols),
                                                   ('Ordinal', ordinal_pipeline ,ordinal_cols ),
                                                   ('Normal', normal_pipeline , normal_cols)

                                                ])
```

**Model Building**

```python
# as continous my target values so base model is Linear Regression

from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_absolute_error , mean_squared_error
lr_model = Pipeline(steps = [('preprocessor', prepocessor),
                             ('regressor' , LinearRegression())
                             ])
lr_model.fit(X_train,y_train)
lr_preds = lr_model.predict(X_test)
lr_preds_training = lr_model.predict(X_train) # for checking the overfitting with testing data , so we have to check the accuracy of training and teating
lr_r2_testing = r2_score(y_test,lr_preds)
lr_r2_training = r2_score(y_train,lr_preds_training)
print(f"Accuracy : {lr_r2_testing}")
print(f"Traing_Accuracy : {lr_r2_training}")
```

```python
# Random Forest Algorithims

from sklearn.ensemble import RandomForestRegressor
rf = Pipeline(steps = [('preprocessor', prepocessor),
                             ('random forest' , RandomForestRegressor(random_state=42))
                             ])

rf.fit(X_train,y_train)
rf_preds =rf.predict(X_test)
rf_preds_training = rf.predict(X_train) # for checking the overfitting with testing data , so we have to check the accuracy of training and teating
rf_r2_testing = r2_score(y_test,rf_preds)
rf_r2_training = r2_score(y_train,rf_preds_training)
print(f"Accuracy : {rf_r2_testing}")
print(f"Training_Accuracy : {rf_r2_training}")

```

```python
# hyperparameter tuning for reducing the overfitting
param_grid = {
          'random forest__n_estimators'     : [100,200,300],
          'random forest__max_depth'       : [5,10,15],
          'random forest__min_samples_split': [2,5,10],
          'random forest__min_samples_leaf' : [1,2,4]

}
from sklearn.model_selection import RandomizedSearchCV
random_search = RandomizedSearchCV(estimator = rf,
                                   param_distributions = param_grid,
                                   n_iter = 15,
                                   cv =5,
                                   scoring = 'r2',
                                   random_state =42,
                                   n_jobs = -1 # use free processor,
                                   )
random_search.fit(X_train,y_train)
rf_best_pipeline = random_search.best_estimator_
rf_best_preds = rf_best_pipeline.predict(X_test)
r2_best_score = r2_score(y_test,rf_best_preds)
print(f"R2 score is {r2_best_score}")
```
